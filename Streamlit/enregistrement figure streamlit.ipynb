```python
def graph_distribution(df,variable):
    df_grouped = df.groupby(variable).size().reset_index(name="Nombre d'incidents")

    fig, ax = plt.subplots(figsize=(8, 6))
    ax.bar(df_grouped[variable], df_grouped["Nombre d'incidents"])
    ax.set_xlabel(variable)
    ax.set_ylabel("Nombre d'incidents")
    ax.set_title(f"Nombre d'incidents pour les valeurs de {variable}")

    filename = "../Donnees/figures/repartion_variable_" + variable
    # Enregistrement de la figure
    fig.savefig(filename)
    plt.close(fig)  # Fermeture de la figure pour libérer la mémoire
```

```python
def creer_pointplot(df, x_col, y_col, title):
    """Crée un point plot avec des barres d'erreur."""
    plt.figure(figsize=(8, 6))  # Ajustez la taille selon vos besoins
    sns.pointplot(x=x_col, y=y_col, data=df)
    plt.title(title)
    plt.xlabel(x_col)
    plt.ylabel(y_col)
    plt.xticks(rotation=45)  # Rotation des étiquettes de l'axe x si nécessaire
    plt.tight_layout()  # Ajuste les paramètres du sous-tracé pour éviter les coupures d'étiquettes

    filename = "../Donnees/figures/evolution_" + x_col + y_col
    # Enregistrement de la figure
    plt.savefig(filename)
    plt.close(plt.gcf())  # Fermeture de la figure actuelle pour libérer la mémoire
```

```python
def creer_evolution(df,x_col,y_col,title):
    plt.figure(figsize=(8, 6))  # Ajustez la taille selon vos besoins
    sns.lineplot(x = x_col,y = y_col, data=df)

    plt.title(title)
    plt.xlabel(x_col)
    plt.ylabel(y_col)

    plt.tight_layout()  # Ajuste les paramètres du sous-tracé pour éviter les coupures d'étiquettes

    filename = "../Donnees/figures/evolution_" + x_col + y_col
    # Enregistrement de la figure
    plt.savefig(filename)
    plt.close(plt.gcf())  # Fermeture de la figure actuelle pour libérer la mémoire
    return plt.gcf()  # Retourne la figure actuelle
```

```python
graph_distribution(df2023,'HourOfCall')
```

```python
fig_pointplot = creer_pointplot(
            df=df2023,
            x_col="HourOfCall",
            y_col="boxcox_TotalResponseTime",
            title="Pointplot de boxcox_TotalResponseTime en fonction de HourOfCall"
        )
```

```python
def creer_evolution2(df, x_col, y_col, title, bins=20):
    """Crée un graphique de l'évolution de la moyenne de y_col en fonction de x_col, avec les données agrégées par intervalles."""

    df['distance_bins'] = pd.cut(df[x_col], bins=bins)  # Créer des intervalles de distance
    df_grouped = df.groupby('distance_bins')[y_col].mean().reset_index()  # Agréger par intervalles

    # Convertir les intervalles en points médians
    df_grouped['distance_bins_mid'] = df_grouped['distance_bins'].apply(lambda interval: interval.mid)

    plt.figure(figsize=(8, 6))
    sns.lineplot(x='distance_bins_mid', y=y_col, data=df_grouped)  # Utiliser les points médians

    plt.title(title)
    plt.xlabel(x_col)
    plt.ylabel(y_col)

    plt.xticks(rotation=45)
    plt.tight_layout()

    filename = "../Donnees/figures/evolution_" + x_col + y_col
    # Enregistrement de la figure
    plt.savefig(filename)
    plt.close(plt.gcf())  # Fermeture de la figure actuelle pour libérer la mémoire
    return plt.gcf()
```

```python
fig = creer_evolution(
    df=df2023,
    x_col = 'distance',
    y_col = "boxcox_TotalResponseTime",
    title = "Evolution du boxcox_TotalResponseTime en fonction de la distance")
```

```python
fig = creer_evolution(
    df=df2023,
    x_col = 'ratioSC',
    y_col = "boxcox_TotalResponseTime",
    title = "Evolution du boxcox_TotalResponseTime en fonction du ration Superficie/Caserne")
```
